> - Lの末尾に18を追加する
> - Lの先頭を "取り出して" 変数xに代入する
> - Lから4を削除する
> - Lに対して[1,2,3]をappend()してみる．その結果がどうなったか，なぜそうなったか，extend()との違いにも触れながら説明してみる．

### 解答欄

**ヒント** : `append()` は引数を「1つの要素」として末尾に追加するので，リスト `[1, 2, 3]` がまるごと1つの要素として入り，
`[6, 8, ..., 18, [1, 2, 3]]` のようなネスト構造になる（長さは1しか増えない）．
一方 `extend()` は引数のリストの「中身を1つずつ」末尾に追加するので，`[6, 8, ..., 18, 1, 2, 3]` となる（長さは3増える）．

```python
L = list(range(2, 17, 2))
L.extend([1, 2, 3])       # extend()なら中身が1つずつ追加される
print('extend([1,2,3]):', L)
print('長さ:', len(L))
```

## リストの中にリストがあるとdeepcopy()が必要

`[[5, 8, 4], 2, 3]` みたいなとき，`copy()` では外側のリストしかコピーされない（内側のリストは共有されたまま）．
`copy` モジュールの `deepcopy()` を使うと中身まで全部コピーされる．「深いコピー」というワードだけ覚えておきましょう．
