## 課題9

> 次のような手順で処理を実行するプログラムを書いてみてください
> - いらっしゃいませと表示する．
> - 販売しているものの「リスト」を紹介する．
> - ユーザーから買いたいものと個数を入力してもらう
> - 店頭にないものの場合は謝って終了する
> - 〇を△個ですね，合計□□円です．のように表示する．
> - お店の在庫は「apple(¥100)」，「lemon(¥150)」，「grape(¥200)」
> - お客が500円を払ったとして，お釣を計算し，〇〇円のお返しですと表示する．足りない場合は足りません，と表示する．

### 解答例

````python
fruits_price = {'apple': 100, 'lemon': 150, 'grape': 200}
payment = 500

print('いらっしゃいませ')
print('商品リスト:', list(fruits_price.keys()))

item = input('何を買いますか?: ')
if item in fruits_price:
    count = int(input('何個買いますか?: '))
    total = fruits_price[item] * count
    print(item, 'を', count, '個ですね, 合計', total, '円です.')

    if payment >= total:
        print(payment - total, '円のお返しです.')
    else:
        print('足りません.')
else:
    print('申し訳ありません,', item, 'は店頭にございません.')
````

> 次のような手順で処理を実行するプログラムを書いてみてください
> - ユーザーが管理者かどうかを問いかける．
> - 管理者ならパスワードを聞き正解なら管理モードへ不正解なら動作を終了する
> - 管理モードでは，追加する商品名と金額を入れてもらう．
> - 商品の追加が終わった後や，ユーザーが管理者でなかった場合は一問目と同じように動作する．

### 解答例

「動作を終了する」を実現するために，買い物に進んでよいかどうかを変数 `shopping` (bool型) に入れておき，最後にそれで分岐している．

````python
fruits_price = {'apple': 100, 'lemon': 150, 'grape': 200}
password = 'denki'
payment = 500

is_admin = input('あなたは管理者ですか? (y/n): ')
if is_admin == 'y':
    user_password = input('パスワードを入力してください: ')
    if user_password == password:
        print('管理モードに入ります.')
        new_name = input('追加する商品の名前: ')
        new_price = int(input('その商品の金額: '))
        fruits_price[new_name] = new_price
        print(new_name, 'を', new_price, '円で追加しました.')
        shopping = True
    else:
        print('パスワードが違います. 終了します.')
        shopping = False
else:
    shopping = True

if shopping:
    print('いらっしゃいませ')
    print('商品リスト:', list(fruits_price.keys()))

    item = input('何を買いますか?: ')
    if item in fruits_price:
        count = int(input('何個買いますか?: '))
        total = fruits_price[item] * count
        print(item, 'を', count, '個ですね, 合計', total, '円です.')

        if payment >= total:
            print(payment - total, '円のお返しです.')
        else:
            print('足りません.')
    else:
        print('申し訳ありません,', item, 'は店頭にございません.')
````

## 課題10

> 課題9の二問目で作成したプログラムに次のような変更を加えてみてください．
> - 管理者のパスワードを3回まで入力できるようにする．
> - 商品名の紹介と同時にそれがいくらなのかも一つづつ表示する．
> - 3種類まで購入できるようにする．（購入点数も，それぞれ選べるように）
> - 支払う金額をユーザーが入力できるようにする（500円で固定ではなく）

### 解答例

- パスワードは `for ... in range(3)` で3回まで繰り返し，正解なら `break`．3回とも間違えたときは for文の `else` で終了処理をする．
- 商品の紹介は辞書を for文で回す（辞書を for で回すとキーが順番に取り出せる）．
- 購入は3回繰り返す for文の中で商品を聞き，途中でやめたいときは `q` を入力すると `break` する．

````python
fruits_price = {'apple': 100, 'lemon': 150, 'grape': 200}
password = 'denki'

shopping = True
is_admin = input('あなたは管理者ですか? (y/n): ')
if is_admin == 'y':
    for i in range(3):                       # パスワードは3回まで
        user_password = input('パスワードを入力してください: ')
        if user_password == password:
            print('管理モードに入ります.')
            new_name = input('追加する商品の名前: ')
            new_price = int(input('その商品の金額: '))
            fruits_price[new_name] = new_price
            print(new_name, 'を', new_price, '円で追加しました.')
            break
        else:
            print('パスワードが違います. (残り', 2 - i, '回)')
    else:                                    # 3回とも間違えたとき
        print('3回間違えたので終了します.')
        shopping = False

if shopping:
    print('いらっしゃいませ')
    print('商品リスト:')
    for name in fruits_price:                # 商品名と値段を一つずつ表示
        print('  ', name, ':', fruits_price[name], '円')

    total = 0
    for i in range(3):                       # 3種類まで購入できる
        item = input('何を買いますか? (やめる場合は q): ')
        if item == 'q':
            break
        if item in fruits_price:
            count = int(input('何個買いますか?: '))
            print(item, 'を', count, '個ですね.')
            total += fruits_price[item] * count
        else:
            print('申し訳ありません,', item, 'は店頭にございません.')

    print('合計', total, '円です.')
    if total > 0:
        payment = int(input('お支払い金額を入力してください: '))
        if payment >= total:
            print(payment - total, '円のお返しです.')
        else:
            print('足りません.')
````

## 課題11

> 課題10の二問目で作成したプログラムに次のような変更を加えてみてください．
> - 管理者のパスワード何度でも入力できるようにする．
> - その代わりユーザーの意思でパスワード入力から抜けられる方法を用意する．
> - 購入できる種類の制限をなくす．（3種類以上も買えるようにする）
> - 購入まで完了したらまた初めからスタートする．

### 解答例

- 全体を `while True:` で囲んで，購入が終わったら最初に戻る．（ノートブック上で止められるように，最初の問いかけで `q` を入力するとプログラムを終了するようにした）
- パスワード入力も `while True:` で何度でも聞き，正解なら管理モードへ，`q` が入力されたらパスワード入力をやめて `break` する．
- 購入も `while True:` で何種類でも聞き，`q` で終了する．

````python
fruits_price = {'apple': 100, 'lemon': 150, 'grape': 200}
password = 'denki'

while True:                                     # 購入が終わったら最初に戻る
    print('=' * 30)
    is_admin = input('あなたは管理者ですか? (y/n, 終了する場合は q): ')
    if is_admin == 'q':
        print('終了します.')
        break

    if is_admin == 'y':
        while True:                             # パスワードは何度でも入力できる
            user_password = input('パスワードを入力してください (やめる場合は q): ')
            if user_password == password:
                print('管理モードに入ります.')
                new_name = input('追加する商品の名前: ')
                new_price = int(input('その商品の金額: '))
                fruits_price[new_name] = new_price
                print(new_name, 'を', new_price, '円で追加しました.')
                break
            elif user_password == 'q':
                print('パスワード入力をやめました.')
                break
            else:
                print('パスワードが違います.')

    print('いらっしゃいませ')
    print('商品リスト:')
    for name in fruits_price:
        print('  ', name, ':', fruits_price[name], '円')

    total = 0
    while True:                                 # 何種類でも購入できる
        item = input('何を買いますか? (やめる場合は q): ')
        if item == 'q':
            break
        if item in fruits_price:
            count = int(input('何個買いますか?: '))
            print(item, 'を', count, '個ですね.')
            total += fruits_price[item] * count
        else:
            print('申し訳ありません,', item, 'は店頭にございません.')

    print('合計', total, '円です.')
    if total > 0:
        payment = int(input('お支払い金額を入力してください: '))
        if payment >= total:
            print(payment - total, '円のお返しです.')
        else:
            print('足りません.')
    print('ありがとうございました.')
````

### 解答例

まず関数を定義する（この時点では何も実行されない）．

````python
PASSWORD = 'denki'


def check_password():
    """パスワード処理をする関数. 正解なら True, ユーザーがやめたら False を返す."""
    while True:
        user_password = input('パスワードを入力してください (やめる場合は q): ')
        if user_password == PASSWORD:
            return True
        elif user_password == 'q':
            print('パスワード入力をやめました.')
            return False
        else:
            print('パスワードが違います.')


def add_product(fruits_price):
    """商品を追加する関数. 追加後の辞書を返す."""
    new_name = input('追加する商品の名前: ')
    new_price = int(input('その商品の金額: '))
    fruits_price[new_name] = new_price
    print(new_name, 'を', new_price, '円で追加しました.')
    return fruits_price


def admin_mode(fruits_price):
    """管理者モード関数. パスワードが正しければ商品を追加し, 辞書を返す."""
    if check_password():
        print('管理モードに入ります.')
        fruits_price = add_product(fruits_price)
    return fruits_price


def show_products(fruits_price):
    """商品を紹介する関数."""
    print('いらっしゃいませ')
    print('商品リスト:')
    for name in fruits_price:
        print('  ', name, ':', fruits_price[name], '円')


def select_items(fruits_price):
    """買うものを選択する関数. {商品名: 個数} の辞書(カート)を返す."""
    cart = {}
    while True:
        item = input('何を買いますか? (やめる場合は q): ')
        if item == 'q':
            break
        if item in fruits_price:
            count = int(input('何個買いますか?: '))
            print(item, 'を', count, '個ですね.')
            cart[item] = count
        else:
            print('申し訳ありません,', item, 'は店頭にございません.')
    return cart


def calc_total(fruits_price, cart):
    """会計を計算する関数. 合計金額を返す."""
    total = 0
    for item in cart:
        total += fruits_price[item] * cart[item]
    return total


def respond_payment(total, payment):
    """支払い額に対して反応する関数."""
    if payment >= total:
        print(payment - total, '円のお返しです.')
    else:
        print('足りません.')


def purchase(fruits_price):
    """お客さんが購入する関数."""
    cart = select_items(fruits_price)
    total = calc_total(fruits_price, cart)
    print('合計', total, '円です.')
    if total > 0:
        payment = int(input('お支払い金額を入力してください: '))
        respond_payment(total, payment)
    print('ありがとうございました.')


def user_mode(fruits_price):
    """購入者モード関数."""
    show_products(fruits_price)
    purchase(fruits_price)
````

メインのループ．（ノートブック上で止められるように `q` で終了できるようにしてある）

````python
fruits_price = {"apple": 100, "lemon": 150, "grape": 200}
while(True):
    permission = input("あなたは管理者ですか?y/n (終了はq)>>>")
    if permission == "q":
        print('終了します.')
        break
    elif permission == "y":
        fruits_price = admin_mode(fruits_price)
    else:
        user_mode(fruits_price)
````
